# Đo độ phủ bổ sung của EasyStock
Danh mục ghi khoảng ngày và số dòng do nguồn công bố. Bảng mẫu kiểm chứng trực tiếp vài mã. Kiểm tra loại tài sản, phiên thiếu và phương pháp điều chỉnh giá trước khi hợp nhất với DNSE. OHLCV chỉ có khung ngày. Notebook tải vào RAM.


```python
from pathlib import Path
import sys
import time

import pandas as pd
from IPython.display import display

cwd = Path.cwd().resolve()
root = next(p for p in (cwd, *cwd.parents) if (p / "ingest/market/easystock/adapter.py").is_file())
sys.path.insert(0, str(root))
from ingest.market.easystock.adapter import catalog, daily

inventory = catalog()
start = pd.Timestamp("2016-01-04")
end = pd.Timestamp("2026-11-17")  # phiên ngày cuối trong lần đo DNSE
spanning = inventory.loc[inventory.first_date.le(start) & inventory.last_date.ge(end)]
print("Mã trong danh mục (mọi loại):", inventory.symbol.nunique())
print("Mã có ngày đầu/cuối bao trùm hai mốc:", spanning.symbol.nunique())
print("Đây là độ phủ theo danh mục, chưa xác nhận đủ từng phiên.")
```

```python
symbols = ["FPT", "HPG", "VCB", "VNINDEX"]
samples = []
for symbol in symbols:
    samples.append(daily(symbol))
    time.sleep(0.25)

df = pd.concat(samples, ignore_index=True)
display(df.groupby("symbol").agg(
    rows=("time", "size"), first_date=("time", "min"), last_date=("time", "max")
))
display(df.head())
```
